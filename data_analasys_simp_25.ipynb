```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore', category=UserWarning, module='urllib3')
```

```python
df: pd.DataFrame = pd.read_json('./Task_2/Task_2.1/Train/train_dataset_posthoc.jsonl', lines=True)
```

```python
from trim_entries import trim_entries, balance_word_count
temp_df: pd.DataFrame = df
temp_df_2: pd.DataFrame = temp_df
temp_df_3: pd.DataFrame = temp_df_2
temp_df_4: pd.DataFrame = temp_df_3
getrimmte_arschhaare = trim_entries(temp_df)
new_word_count = balance_word_count(temp_df_2)

word_count_before_trim_df = trim_entries(balance_word_count(temp_df_3))
trim_before_word_count_df = balance_word_count(trim_entries(temp_df_4))
```

## Nr. of spurious sentences

```python
print(
    f'Percentage of spurious entries (before): {df['is_spurious'].mean():.4%}'
)
print(
    f'Percentage of spurious entries (after just trim): {getrimmte_arschhaare['is_spurious'].mean():.4%}'
)
print(
    f'Percentage of spurious entries (after just word count): {new_word_count['is_spurious'].mean():.4%}'
)
print(
    f'Percentage of spurious entries (word count before trim): {word_count_before_trim_df['is_spurious'].mean():.4%}'
)
print(
    f'Percentage of spurious entries (trim before word count): {trim_before_word_count_df['is_spurious'].mean():.4%}'
)
```

## Avg. sentence length grouped by `is_spurious`

```python
print('-----before-----')
df['word_count'] = df['sentence'].str.findall(r'[\w-]+').str.len()

print(
    f'Mean:\n{df.groupby('is_spurious')['word_count'].mean()}\n'
)
print(
    f'Median:\n{df.groupby('is_spurious')['word_count'].median()}'
)

print('-----spurious-calculation-after-----')
getrimmte_arschhaare['word_count'] = getrimmte_arschhaare['sentence'].str.findall(r'[\w-]+').str.len()

print(
    f'Mean:\n{getrimmte_arschhaare.groupby('is_spurious')['word_count'].mean()}\n'
)
print(
    f'Median:\n{getrimmte_arschhaare.groupby('is_spurious')['word_count'].median()}'
)

print('-----word-count-balance-after-----')
new_word_count['word_count'] = new_word_count['sentence'].str.findall(r'[\w-]+').str.len()

print(
    f'Mean:\n{new_word_count.groupby('is_spurious')['word_count'].mean()}\n'
)
print(
    f'Median:\n{new_word_count.groupby('is_spurious')['word_count'].median()}'
)

print('-----word_count_before_trim-----')
word_count_before_trim_df['word_count'] = word_count_before_trim_df['sentence'].str.findall(r'[\w-]+').str.len()

print(
    f'Mean:\n{word_count_before_trim_df.groupby('is_spurious')['word_count'].mean()}\n'
)
print(
    f'Median:\n{word_count_before_trim_df.groupby('is_spurious')['word_count'].median()}'
)

print('-----trim_before_word_count-----')
trim_before_word_count_df['word_count'] = trim_before_word_count_df['sentence'].str.findall(r'[\w-]+').str.len()

print(
    f'Mean:\n{trim_before_word_count_df.groupby('is_spurious')['word_count'].mean()}\n'
)
print(
    f'Median:\n{trim_before_word_count_df.groupby('is_spurious')['word_count'].median()}'
)
```

## Nr. of single character sentences

```python
print('-----before-----')
df['is_single_character'] = df['sentence'].str.len() == 1

print(
    f'Mean:\n{df.groupby('is_spurious')['is_single_character'].mean()}'
)

print('-----spurious-calculation-after-----')
getrimmte_arschhaare['is_single_character'] = getrimmte_arschhaare['sentence'].str.len() == 1

print(
    f'Mean:\n{getrimmte_arschhaare.groupby('is_spurious')['is_single_character'].mean()}'
)

print('-----word-count-after-----')
new_word_count['is_single_character'] = new_word_count['sentence'].str.len() == 1

print(
    f'Mean:\n{new_word_count.groupby('is_spurious')['is_single_character'].mean()}'
)

print('-----word-count-before-trim-----')
word_count_before_trim_df['is_single_character'] = word_count_before_trim_df['sentence'].str.len() == 1

print(
    f'Mean:\n{word_count_before_trim_df.groupby('is_spurious')['is_single_character'].mean()}'
)

print('-----trim-before-word-count-----')
trim_before_word_count_df['is_single_character'] = trim_before_word_count_df['sentence'].str.len() == 1

print(
    f'Mean:\n{trim_before_word_count_df.groupby('is_spurious')['is_single_character'].mean()}'
)
```

## Percentage of labels present

```python
description_width_1 = 75
description_width_2 = 40

labels = [
    'No error',
    'A1. Random generation',
    'A2. Syntax error', 'A3. Contradiction',
    'A4. Simple punctuation / grammar errors',
    'A5. Redundancy',
    'B1. Format misalignement',
    'B2. Prompt misalignement',
    'C1. Factuality hallucination',
    'C2. Faithfulness hallucination',
    'C3. Topic shift',
    'D1.1. Overgeneralization',
    'D1.2 Overspecification of Concepts',
    'D2.1. Loss of Informative Content',
    'D2.2. Out-of-Scope Generation'
]

no_errors_before = df2[labels[0]].mean()
random_generation_before = df2[labels[1]].mean()
syntax_error_before = df2[labels[2]].mean()
contradiction_before = df2[labels[3]].mean()
punctuation_grammar_before = df2[labels[4]].mean()
redundancy_before = df2[labels[5]].mean()
format_misalignment_before = df2[labels[6]].mean()
prompt_misalignment_before = df2[labels[7]].mean()
factuality_hallucination_before = df2[labels[8]].mean()
faithfulness_hallucination_before = df2[labels[9]].mean()
topic_shift_before = df2[labels[10]].mean()
overgeneralization_before = df2[labels[11]].mean()
overspecification_before = df2[labels[12]].mean()
loss_of_info_content_before = df2[labels[13]].mean()
out_of_scope_generation_before = df2[labels[14]].mean()

group_A_total_before = random_generation_before + syntax_error_before + contradiction_before + punctuation_grammar_before + redundancy_before
group_B_total_before = format_misalignment_before + prompt_misalignment_before
group_C_total_before = factuality_hallucination_before + faithfulness_hallucination_before + topic_shift_before
group_D_total_before = overgeneralization_before + overspecification_before + loss_of_info_content_before + out_of_scope_generation_before

print(
    f'{'Entries with no errors (before):':<{description_width_1}} {no_errors_before:.4%}\n'
    f'{f'Entries classified as {labels[1]} (before):':<{description_width_1}} {random_generation_before:.4%}\n'
    f'{f'Entries classified as {labels[2]} (before):':<{description_width_1}} {syntax_error_before:.4%}\n'
    f'{f'Entries classified as {labels[3]} (before):':<{description_width_1}} {contradiction_before:.4%}\n'
    f'{f'Entries classified as {labels[4]} (before):':<{description_width_1}} {punctuation_grammar_before:.4%}\n'
    f'{f'Entries classified as {labels[5]} (before):':<{description_width_1}} {redundancy_before:.4%}\n'
    f'{f'Entries classified as {labels[6]} (before):':<{description_width_1}} {format_misalignment_before:.4%}\n'
    f'{f'Entries classified as {labels[7]} (before):':<{description_width_1}} {prompt_misalignment_before:.4%}\n'
    f'{f'Entries classified as {labels[8]} (before):':<{description_width_1}} {factuality_hallucination_before:.4%}\n'
    f'{f'Entries classified as {labels[9]} (before):':<{description_width_1}} {faithfulness_hallucination_before:.4%}\n'
    f'{f'Entries classified as {labels[10]} (before):':<{description_width_1}} {topic_shift_before:.4%}\n'
    f'{f'Entries classified as {labels[11]} (before):':<{description_width_1}} {overgeneralization_before:.4%}\n'
    f'{f'Entries classified as {labels[12]} (before):':<{description_width_1}} {overspecification_before:.4%}\n'
    f'{f'Entries classified as {labels[13]} (before):':<{description_width_1}} {loss_of_info_content_before:.4%}\n'
    f'{f'Entries classified as {labels[14]} (before):':<{description_width_1}} {out_of_scope_generation_before:.4%}\n'
)
print(
    f'{f'Total entries in Group {'Fluency'}':<{description_width_2}} {group_A_total_before:.4%}\n'
    f'{f'Total entries in Group {'Alignment'}':<{description_width_2}} {group_B_total_before:.4%}\n'
    f'{f'Total entries in Group {'Information'}':<{description_width_2}} {group_C_total_before:.4%}\n'
    f'{f'Total entries in Group {'Simplification'}':<{description_width_2}} {group_D_total_before:.4%}'
)
```

# The plot thickens

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Label distribution

```python
df2['word_count'] = df2['source sentence'].str.findall(r'[\w-]+').str.len()

df2_long = df2.melt(id_vars=['word_count'], value_vars=labels, var_name='label_type', value_name='is_active')
df2_plot_data = df2_long[df2_long['is_active'] == True].copy()

heatmap_data = (
    df2_plot_data
    .groupby(['word_count', 'label_type'])
    .size()
    .unstack(fill_value=0)
)
heatmap_data_norm = heatmap_data.div(heatmap_data.sum(axis=1), axis=0)

plt.figure(figsize=(12,6))
ax = sns.heatmap(
    heatmap_data_norm,
    cmap='viridis'
)
plt.xlabel('Label')
plt.ylabel('Word Count')
plt.show()


df_counts = (
    df2_plot_data
    .groupby(['word_count', 'label_type'])
    .size()
    .reset_index(name='count')
)
df_counts['prob'] = df_counts.groupby('word_count')['count'].transform(lambda x: x / x.sum())

plt.figure(figsize=(12,6))
ax2 = sns.lineplot(
    data=df_counts,
    x='word_count',
    y='prob',
    hue='label_type'
)
plt.xlabel('Word Count')
plt.ylabel('Probability')
sns.move_legend(ax2, "upper left", bbox_to_anchor=(1, 1))
plt.show()
```
